# ROCProfiler-Compute Jupyter Analysis Example

This notebook demonstrates the Jupyter API for rocprofiler-compute.

## Prerequisites

1. Profiling data from `rocprof-compute profile` (or equivalent).
2. Path to the workload directory.

## Setup and import

```python
import sys
from pathlib import Path

# Add src to path (adjust if running from another location)
src_path = Path("../src").resolve()
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

import rocprof_compute_jupyter as rc
```

## Load performance data

Replace `perf_data_dir` with your workload path.

```python
perf_data_dir = "/path/to/your/workload_dir"

rc.open(perf_data_dir)
```

## Display analysis results

```python
rc.analysis()
```
